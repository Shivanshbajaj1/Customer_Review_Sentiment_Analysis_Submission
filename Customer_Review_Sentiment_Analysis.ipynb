### Objective
Build a simple NLP + Machine Learning system that predicts whether a customer review is **Positive, Negative, or Neutral**.

### Workflow
**Customer Reviews → Text Cleaning → TF-IDF Features → Logistic Regression → Sentiment Prediction → Evaluation**

This notebook follows the assignment requirements: small dataset, basic preprocessing, TF-IDF, one classification model, accuracy evaluation, and testing on at least three new reviews.

## 1. Import Libraries

We use:
- **Pandas** for dataset handling
- **Regular Expressions (re)** for basic text preprocessing
- **Scikit-learn** for train/test split, TF-IDF, Logistic Regression, and evaluation

```python
import pandas as pd
import re

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print("Libraries imported successfully.")
```

## 2. Load the Dataset

The assignment suggests creating a small CSV file with two columns:
`review` and `sentiment`.

The dataset used here contains **30 reviews**: 10 Positive, 10 Negative, and 10 Neutral.

```python
# Load the CSV file
df = pd.read_csv("customer_reviews.csv")

print("Dataset shape:", df.shape)
display(df.head(5))
```

## 3. Check the Sentiment Distribution

```python
print(df["sentiment"].value_counts())
```

## 4. Basic Text Preprocessing

The review text is cleaned by:
1. converting text to lowercase,
2. removing non-letter characters,
3. removing extra spaces.

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df["clean_review"] = df["review"].apply(clean_text)

display(df[["review", "clean_review"]].head(5))
```

## 5. Split Data into Training and Testing Sets

```python
X = df["clean_review"]
y = df["sentiment"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 6. Convert Text into TF-IDF Features

**TF-IDF (Term Frequency-Inverse Document Frequency)** converts text into numerical values. Words that are useful for distinguishing documents receive more importance than words that occur everywhere.

```python
vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("TF-IDF training matrix shape:", X_train_tfidf.shape)
print("TF-IDF testing matrix shape:", X_test_tfidf.shape)
```

## 7. Train the Machine Learning Model

### Model: Logistic Regression

Logistic Regression is a simple classification algorithm suitable for text classification when the text has been converted into numerical features such as TF-IDF.

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_tfidf, y_train)

print("Model training completed.")
```

## 8. Predict Sentiment for the Test Data

```python
y_pred = model.predict(X_test_tfidf)

results = pd.DataFrame({
    "Customer Review": X_test,
    "Actual Sentiment": y_test,
    "Predicted Sentiment": y_pred
})

display(results)
```

## 9. Calculate Accuracy

```python
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy * 100:.2f}%")
```

## 10. Confusion Matrix

```python
labels = ["Negative", "Neutral", "Positive"]

cm = confusion_matrix(y_test, y_pred, labels=labels)

cm_df = pd.DataFrame(
    cm,
    index=[f"Actual {label}" for label in labels],
    columns=[f"Predicted {label}" for label in labels]
)

display(cm_df)
```

## 11. Classification Report

```python
print(classification_report(y_test, y_pred, zero_division=0))
```

## 12. Test the Model on 3 New Reviews

The assignment requires at least three new reviews written by the student.

```python
new_reviews = [
    "Amazing product and fast delivery!",
    "The quality is terrible and disappointing.",
    "The package arrived today."
]

new_clean = [clean_text(review) for review in new_reviews]
new_features = vectorizer.transform(new_clean)
new_predictions = model.predict(new_features)

new_results = pd.DataFrame({
    "Customer Review": new_reviews,
    "Predicted Sentiment": new_predictions
})

display(new_results)
```
